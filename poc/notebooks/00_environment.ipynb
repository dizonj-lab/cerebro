## Experimental Principles

CEREBRO follows an evidence-driven development approach.

Every experiment must define:

1. Hypothesis
2. Input dataset
3. Ground truth
4. Processing method
5. Model/tool and version
6. Configuration
7. Evaluation metrics
8. Success criteria
9. Results
10. Conclusion
11. Reproducibility information

No architecture or model selection should be accepted solely because
the output appears reasonable.

Decisions must be supported by measurable evidence where practical.

## Experiment Identification

Experiment ID format:

EXP-<DOMAIN>-<NUMBER>

Examples:

EXP-OCR-001
EXP-ASR-001
EXP-CHUNK-001
EXP-EMBED-001
EXP-RETRIEVAL-001
EXP-FUSION-001

Each experiment must reference:

- Dataset version
- Pipeline version
- Model/tool version
- Configuration
- Ground truth
- Metrics
- Result
- Related ADR, if a decision is made

## CEREBRO Integrity Chain

Every derived piece of knowledge must remain traceable through:

Source Artifact
→ Extraction
→ Transformation
→ Normalization
→ Knowledge Fragment
→ Embedding / Entity / Relationship
→ Retrieval
→ Reasoning
→ Recollection

Each stage should preserve, where applicable:

- Source identity
- Source location
- Transformation method
- Model/tool
- Model/tool version
- Timestamp
- Configuration
- Quality measurement
- Parent representation

## Retrieval Evaluation

Retrieval quality must be evaluated independently from extraction quality.

Candidate metrics:

- Recall@K
- Precision@K
- Hit Rate@K
- Mean Reciprocal Rank (MRR)
- Retrieval latency
- Provenance completeness

Retrieval experiments must use controlled ground-truth queries.
