## EXP-GT-002 — Ground-Truth Annotation Model

### Objective

Define the manually verified annotation structure used by
CEREBRO-GT-v0.1 to evaluate future knowledge-extraction models.

The benchmark distinguishes between:

1. **Explicit source truth** — directly supported by the artifact.
2. **Human annotation** — manually assigned semantic labels.
3. **Derived inference** — conclusions produced later by AI or reasoning.

Derived inference must never be represented as source truth.

``` mermaid
flowchart LR
    A[Artifact] --> KF[Knowledge Fragment]
    KF --> E[Entities]
    KF --> C[Concepts]
    KF --> R[Relationships]
    KF --> T[Temporal Facts]
    E --> P[Source Provenance]
    C --> P
    R --> P
    T --> P
```

### Simple annotation validator

````python
def validate_annotation(annotation):

    required = [
        "annotation_id",
        "dataset",
        "artifact_id",
        "fragment_id",
        "entities",
        "concepts",
        "relationships",
        "temporal_facts",
        "annotation_method",
        "manually_verified"
    ]

    missing = [
        field
        for field in required
        if field not in annotation
    ]

    return {
        "valid": len(missing) == 0,
        "missing": missing
    }


result = validate_annotation(annotation_001)

print(result)

assert result["valid"]

print("✓ Annotation schema valid")
````

````python
import json
from pathlib import Path

benchmark = {
    "benchmark": "CEREBRO-GT-v0.1",
    "artifact": artifact,
    "fragment": knowledge_fragment,
    "annotation": annotation
}

output_dir = Path("../data/ground_truth")
output_dir.mkdir(parents=True, exist_ok=True)

benchmark_path = output_dir / "CEREBRO-GT-v0.1.json"

with benchmark_path.open("w", encoding="utf-8") as f:
    json.dump(benchmark, f, indent=2, ensure_ascii=False)

print("✓ Benchmark persisted")
print("Path:", benchmark_path)
````
