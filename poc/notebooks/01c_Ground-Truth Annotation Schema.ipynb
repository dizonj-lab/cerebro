## EXP-GT-002 — Ground-Truth Annotation Model

### Objective

Define the manually verified annotation structure used by
CEREBRO-GT-v0.1 to evaluate future knowledge-extraction models.

The benchmark distinguishes between:

1. **Explicit source truth** — directly supported by the artifact.
2. **Human annotation** — manually assigned semantic labels.
3. **Derived inference** — conclusions produced later by AI or reasoning.

Derived inference must never be represented as source truth.

``` mermaid
flowchart LR
    A[Artifact] --> KF[Knowledge Fragment]
    KF --> E[Entities]
    KF --> C[Concepts]
    KF --> R[Relationships]
    KF --> T[Temporal Facts]
    E --> P[Source Provenance]
    C --> P
    R --> P
    T --> P
```

### Define the three evidence levels

````python
EVIDENCE_LEVELS = {
    "explicit": {
        "description": "Directly stated or observable in the source."
    },
    "annotated": {
        "description": "Human-assigned interpretation grounded in the source."
    },
    "inferred": {
        "description": "Derived conclusion requiring reasoning beyond the explicit source."
    }
}

EVIDENCE_LEVELS
````

### Assemble the annotation record

````python
annotation_001 = {
    "annotation_id": "ANN-0001",

    "dataset": "CEREBRO-GT-v0.1",

    "artifact_id": "ART-0001",
    "fragment_id": "KF-0001",

    "entities": ground_truth_entities_001,
    "concepts": ground_truth_concepts_001,
    "relationships": ground_truth_relationships_001,
    "temporal_facts": ground_truth_temporal_001,

    "annotation_method": "manual",
    "manually_verified": True
}

annotation_001
````

### Add a simple annotation validator

````python
def validate_annotation(annotation):

    required = [
        "annotation_id",
        "dataset",
        "artifact_id",
        "fragment_id",
        "entities",
        "concepts",
        "relationships",
        "temporal_facts",
        "annotation_method",
        "manually_verified"
    ]

    missing = [
        field
        for field in required
        if field not in annotation
    ]

    return {
        "valid": len(missing) == 0,
        "missing": missing
    }


result = validate_annotation(annotation_001)

print(result)

assert result["valid"]

print("✓ Annotation schema valid")
````
